### Step Two: Adding a Legend

To add a legend, we:
- Add a ‘label’ argument to each plot object
- Call ‘legend’ and specify our legend location

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1" )
ax.plot(x,y2, label = "Person 2" )
ax.legend(loc='lower right')
fig.show()
```

#### Modifying our legend

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='lower right',
frameon = True, #add frame around the legend
fontsize = 12, #change font size
ncol = 2, #specify number of columns
shadow = True )
fig.show()
```

#### Moving the legend outside of the plot area
- We can position the legend outside of our plot by
- Specifying legend location as before
- Specifying where we want to anchor that location on our figure using bbox_to_anchor

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.plot(x, y1, label="Person 1")
ax.plot(x, y2, label="Person 2")

ax.legend(
    loc="upper left",
    bbox_to_anchor=(1, 1)
)

plt.show()
```

## Step Two: Text and Annotations

```python
### Let’s start by putting our data into a scatter plot this time
fig, ax = plt.subplots(figsize=(5, 3))

ax.scatter(x, y1, label="Person 1")  # scatter plot for first dataset
ax.scatter(x, y2, label="Person 2")  # scatter plot for second dataset

ax.legend(loc='lower right')  # add legend in bottom-right corner

### Then use the text() function to specify coordinates and text
ax.text(
    10, 95, "This value is important!"  # x, y position and text
)

plt.show()  # display the figure (use this instead of fig.show() in Jupyter)
```

### Modifying text annotations

```python
ax.text(
    10, 95, "This value is important!",
    ha='center',   # horizontal alignment
    color='red',   # font colour
    size=20        # font size
)

plt.show()
```

### Positioning text annotations

We can do this using transforms
- Start by making a simple, empty set of axes

```python
fig, ax = plt.subplots()

# Set axis limits: [xmin, xmax, ymin, ymax]
ax.axis([0, 10, 0, 10])

# Add text using DATA coordinates (x=1, y=5 in data space)
ax.text(
    1, 5, "Data: (1, 5)",
    transform=ax.transData
)

plt.show()  # display the figure in Jupyter / VS Code
```

- Then we can position our annotation with reference to the data (like before), the axes,
or the figure (referencing axis/figure dimensions)

```python
fig, ax = plt.subplots()

# Set axis limits so data coordinates are clear
ax.axis([0, 10, 0, 10])

# DATA coordinates: tied to the data values on the axes
ax.text(
    1, 5, "Data: (1, 5)",
    transform=ax.transData
)

# AXES coordinates: (0,0) bottom-left, (1,1) top-right of the axes
ax.text(
    0.5, 0.1, "Axes: (0.5, 0.1)",
    transform=ax.transAxes
)

# FIGURE coordinates: relative to the entire figure window
ax.text(
    0.2, 0.2, "Figure: (0.2, 0.2)",
    transform=fig.transFigure
)

plt.show()  # display the figure
```

### Annotating with arrows
- We can use the annotate() function to annotate with arrows and text (e.g. to focus at a
specific point)
◦ We have to specify the location to point at and the location where we want our text

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate('This is important!', xy=(10, 95), xytext=(20, 94),
arrowprops=dict(facecolor='black'))
fig.show()
```

### Modifying arrows

```python
fig, ax = plt.subplots(figsize=(5, 3))

# Scatter plots
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")

# Legend
ax.legend(loc='lower right')

# Add an annotation with an arrow pointing to a specific data point
ax.annotate(
    'This is important!',          # text to display
    xy=(10, 95),                   # point being annotated (data coordinates)
    xytext=(20, 94),               # location of the text
    arrowprops=dict(
        arrowstyle="<->",        # style of the arrow
        color="hotpink"            # arrow colour
    )
)

plt.show()  # display the figure
```

## Step Three: Axis Labels

## Removing tick marks and labels
- We can remove tick marks or labels from our axis like so:
- Use locator to remove ticks and labels
- Use formatter to remove only labels

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.yaxis.set_major_locator(plt.NullLocator())
ax.xaxis.set_major_formatter(plt.NullFormatter())
```

### Limiting tick marks

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MaxNLocator(3))
```

### Tick mark intervals

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
```

### Rotating Axis Labels

```python
fig, ax = plt.subplots(figsize=(5, 3))

# Scatter plots
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")

# Legend in lower-right corner
ax.legend(loc='lower right')  # straight quotes required

# Rotate x-axis tick labels
plt.xticks(rotation=45, ha='right')

plt.show()
```

## Activity 2: Modifying axis title fonts

Can you use what we did in a previous lesson to modify our x axis title?
- We want a serif font with indigo colour, and for the axis title to be “Shiny New X Axis!”
- Use plt.xlabel()

HINT: Recall fontdict

```python
import matplotlib.pyplot as plt

# Define font style for the axis label
font1 = {
    'family': 'serif',   # font family
    'color': 'indigo',   # font color
    'weight': 'normal',  # optional: normal/bold
    'size': 18           # font size
}

fig, ax = plt.subplots(figsize=(5, 3))

# Scatter plots
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")

# Legend in lower-right corner
ax.legend(loc='lower right')

# X-axis label with custom font
plt.xlabel('Shiny New X Axis!', fontdict=font1)

plt.show()
```
