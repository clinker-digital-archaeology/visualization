## Activity 1: Making a a Basic Figure with matplotlib

### Step 1: making sample data so we have something to plot:


```python
np.random.seed(613)
x = np.arange(50)
y=np.random.randint(0,100,50)
```

Notes: 
- We create data so we have something to visualize.
- np.random.seed() ensures our random values are reproducible.
- x represents an ordered sequence (e.g., time or index).
- y represents random values that will be plotted against x.

### Step 2: then make our basic scatterplot (breakdown coming later)

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y)
```

Notes: 
- plt.subplots() creates a figure and an axis object.
- The axis object (ax) is where all plotting happens.
- ax.scatter() creates a scatterplot showing the relationship between x and y.

### Step 3: Modifying the graph 
- Keep the figure and axis definition the same
- Only edit our axis object

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax. bar (x, y)
```

Notes:
- We keep the same figure and axis structure.
- Only the plotting method changes (scatter → bar).
- This highlights how different plot types can visualize the same data in different ways.

### Step 4: Making line plot 

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
```

Notes:
- ax.plot() connects data points with a line.
- Line plots are commonly used to show trends or changes over time.

### Step 1: Create Histogram

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y)
```

NOTE: 
- Only y is required in ax.hist() because the x-axis is created automatically from the binned values of y, and the y-axis represents the frequency of those values.
- A histogram shows the distribution of a single variable.
- The y-axis represents the frequency (count) of values in each bin.

### Step 2: Adding Plot Label

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax. set_title( 'Total growth over time')
ax.set_ylabel('Total growth')
ax. set_xlabel ('Years since start')
fig. tight_layout ()
```

Notes:
- Titles and axis labels provide context and meaning to a visualization.
- tight_layout() helps prevent text from overlapping or being cut off.

### Step 3: Modifying labels and titles
- Modify our labels and titles' appearance by using font dictionary
- First, make a dictionary with the properties we want for our labels


```python
font1 = {'family': 'sans-serif', 'color' : 'blue' , 'size': 20}
font2 = {'family': 'monospace', 'color' : 'green' , 'size' : 14}
```

Notes:
- Font dictionaries allow customization of text appearance.
- Different fonts, sizes, and colors can be used to establish hierarchy and improve readability.

### Step 4: Then we'll add fontdict arguments to our previous lineplot

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax. plot (x,y)
ax. set_title('Total growth over time', fontdict = font1)
ax. set_ylabel ('Total growth', fontdict = font2)
ax. set_xlabel('Years since start', fontdict = font2)
fig.tight_layout ()
```

### Step 4: Modifying data points

- Remake our basic scatterplot, but this time add color and marker (argument is in American spelling)


```python
fig, ax=plt. subplots (figsize=(5,3))
ax. scatter(x,y, marker='*', color="indigo")
plt.show
```

Notes:
- Marker shape and color help distinguish data visually.
- Arguments such as marker and color use American spelling.
- Scatterplots emphasize individual observations.

### Step 5: We can also adjust the style and width of the line connecting our points

```python
fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,y, marker='*', color='indigo', linestyle='--')
```

Notes:
- Linestyle controls the appearance of the line (solid, dashed, etc.).
- Line styles can help differentiate multiple datasets in the same figure.

### Step 6: We can modify colour using named colors
See: https://matplotlib.org/stable/gallery/color/named_colors.html##sphx-glr-gallery- color-named-colors-py

Or, we can use hex codes

```python
fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,y, marker='*', color='indigo', linestyle='--')

# color being the new addition
ax.plot (x,y,marker='*', color = '#7425b9', linestyle = '--', linewidth = 2)
fig.show()
```

### Step 6 (continued): Modifying the graph further

```python
fig,ax=plt.subplots(figsize=(5,3))

ax.plot(
    x, y,
    marker='*',
    markersize=12,
    color='indigo',
    linestyle='--',
    linewidth=2,
    markerfacecolor='#000000',
    markeredgecolor='#fa9359'
)
```

```python
# Chosen a Density plot of several variables from the gallery 

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from scipy.stats import gaussian_kde

df = sns.load_dataset('iris')

fig, ax = plt.subplots(figsize=(6, 4))


sepal_width = df['sepal_width']
sepal_length = df['sepal_length']


x_width = np.linspace(sepal_width.min(), sepal_width.max(), 500)
x_length = np.linspace(sepal_length.min(), sepal_length.max(), 500)

kde_width = gaussian_kde(sepal_width)
kde_length = gaussian_kde(sepal_length)

ax.plot(x_width, kde_width(x_width), color='red', linewidth=2)
ax.fill_between(x_width, kde_width(x_width), color='red', alpha=0.4)

ax.plot(x_length, kde_length(x_length), color='blue', linewidth=2)
ax.fill_between(x_length, kde_length(x_length), color='blue', alpha=0.4)

ax.grid(True, linestyle='--', alpha=0.6)
ax.set_xlabel('sepal_width')
ax.set_ylabel('Density')
ax.set_title('KDE Plot (Session 1 Activity)')

plt.show()

```
